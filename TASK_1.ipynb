```python
import pandas as pd
import numpy as np
data = {
    'ID': [101, 102, 103, 103, 104, 105, 106, 106],
    'Name': ['Alice', 'Bob', None, 'Charlie', 'David', 'Eva', 'Frank', 'Frank'],
    'Date': ['10/12/23', '2023-12-10', 'Dec 12, 2023', '12-10-2023',
             None, '2023/12/11', '11-12-23', '11-12-23'],
    'Salary': [50000, 60000, None, 70000, 80000, 90000, None, None]
}
df = pd.DataFrame(data)
print("RAW DATASET")
print(df)
df = df.drop_duplicates()
print("\nAfter Removing Duplicates:")
print(df)
df['Name'] = df['Name'].fillna('Unknown')
df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
df['Date'] = df['Date'].fillna(method='ffill')
print("\nAfter Handling Missing Values:")
print(df)
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df['Date'] = df['Date'].dt.strftime('%Y-%m-%d')
print("\nAfter Standardizing Dates:")
print(df)
print("\nMissing Values Count:")
print(df.isnull().sum())
print("\nUnique IDs:", df['ID'].nunique())
print("\nSalary Statistics:")
print(df['Salary'].describe())
df.to_csv('cleaned_dataset.csv', index=False)
print("\nCleaned dataset saved as 'cleaned_dataset.csv'")
print("\nFINAL CLEANED DATASET")
print(df)
```
